```python
# String Methods(Functions)
v2="vishw sarvaiya"

# 1. lower,upper
print(v2.lower())
print(v2.upper())

#2. capitalize,title
print(v2.capitalize())#make the letter capitle which is 1st in sentence
print(v2.title())#make the letter capital as the space arrives{makes wor to word capital letter}

#3. swapcase
print(v2.swapcase())#swap the cases as lower to upper or vice versa

#4. count
print(v2.count("w"))#gives the number how many time the letter is observed

#5. find,rfind
print(v2.find("shw"))#gives the index from where the substring starts
print(v2.find("shw",5,len(v2)))#gives -1 cause substring is not found in given range
print(v2.rfind("a"))
#same works for index,rindex but if not found then it gives error

#5. startswith,endswith
print(v2.startswith("vishw"))
print(v2.endswith("ya"))

#6. isalnum,isalpha
al="vis234"
alp="vishhhuu"
print("for string v2",v2.isalnum())#this gives true only if the word is of alphabets and number not even space
print("for string al",al.isalnum())
print("for string alp",alp.isalpha())#this must contain only alphabets not even space then only true

#7. islower,isupper
print(al.islower())
o="V23"
print(o.isupper())

#8. isnumeric or isdigit
n1="12345"
print(n1.isnumeric())
print(n1.isdigit())

#9. isspace
F=" "
print("if the string only contain space then true ",F.isspace())
```

```python
#WAP to find no. of consonants, no. of vowels , no. of special characters, no.of words 
#sum of digit given in string enterd by user

import string
stra=input("Enter the Sentence :")
strvow=0
strcon=0
strspec=0
strwrds=1
sm=0
for i in stra.lower():
    if i.isalpha():
        if i in 'aeiou':
            strvow+=1
        else:
            strcon+=1
    elif i.isspace():
        strwrds+=1
    elif i.isnumeric():
        sm=sm+int(i)
    elif i in string.punctuation:
        strspec+=1
    else:
        print("Invalid..")
print("no. of vowels ",strvow)
print("no. of consonants ",strcon)
print("no. of words ",strwrds)
print("no. of special characters ",strspec)
print("sum of numbers ",sm)
```

```python
#WAP to find max. even number max. odd no. and min. even&odd from tuple of numbers enterd by user
maxe=None
mine=None
maxo=None
mino=None
tup=eval(input("Enter Tuple Of Numbers:- "))
for i,j in enumerate(tup):
    if j%2==0:
        if maxe is None or j>maxe:
            maxe=j
        if mine is None or j<mine:
            mine=j
    elif j%2!=0:
        if maxo is None or j>maxo:
            maxo=j
        if mino is None or j<mino:
            mino=j
    else:
        print("Invalid input")
print("Max even:",maxe,"| Min even:",mine,"| Max odd:",maxo,"| Max odd:",mino)

```

```python
#by concating tuples

tup1=eval(input("Enter Tuple Of Numbers:- "))
t_even=()
t_odd=()
for i in tup1:
    if i%2==0:
        t_even+=(i,)
    else:
        t_odd+=(i,)
maxe=max(t_even)
mine=min(t_even)
maxo=max(t_odd)
mino=min(t_odd)       
print("Max even:",maxe,"| Min even:",mine,"| Max odd:",maxo,"| Max odd:",mino)
```
